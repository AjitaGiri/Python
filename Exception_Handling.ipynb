```python
def factorial(n):
  assert n>=0, "The factorial of negative integer is not defined"
  if n==0:
    return 1
  else:
    return n*factorial(n-1)

print(factorial(5))
```

```python
#Custom Exception
#User defined Exception

class   NegativeNumberError(Exception):
  def __init__(self,value):
    self.value=value
    self.message=f'Factorial is not defined for negative numbers: {value}'
    super().__init__(self.message)

def factorial(n):
  if n<0:
    raise NegativeNumberError(n)
  if n==0:
   return 1
  else:
    return n* factorial(n-1)

try:
  print(factorial(6))
  print(factorial(-6))
except NegativeNumberError as e:
  print("Caught an exception:",e)
```
