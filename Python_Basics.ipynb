#Types of arguments

1. arbitary arguments(*args):  
allows us to pass variable number of positional arguments

2. keyword arguments: pass argument in key-value pair


3. arbitary keyword arguments(**kwargs):  
allows us  to pass variable number of keyword arguments


4. default arguments

```python
#arbitary args
def add(*args):
    return sum(args)

print(add(4,5))
```

```python
#greeting multiple people
def greet(*args):
  for name in args:
    print(f'Namaste {name}')

greet('Ram','Jake')
```

```python
#keyword argument

def my_func(name,band):
  print(f'Hey my name is {name} and am from {band}')

my_func(name='Jake',band='Enhypen')

```

```python
#arbitary keyword arguments (**kwargs)
def print_kwargs(**kwargs):
  print(kwargs)

print_kwargs(song1='Daydream', song2='No doubt')
```

```python
#loop through kwargs
def student_info(**kwargs):
  for key,value in kwargs.items():
    print(f'{key}:{value}')

student_info(name='Ajita', age='25')
```

```python
#default arguments
def my_funct(country="USA"):
  print(f'I am from {country}')

my_funct()
my_funct("Nepal")
```

```python
#check prime
#two factors: 1 and number itself
def check_prime(n):
  if n<=1:
    return False
  for i in range(2,n):
    if n%i==0:
      #print("Not Prime")
      return False
    else:
      #print("Prime")
      return True

check_prime(6)
```

#Recursion:  
function that calls itself

to stop calling we must specify the base case

**skeleton**

def function_name(parameters):

  if base condition:

    return some value

  else:

   return function_name(smaller problem)

```python
#factorial of a number using recursion
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)

num=int(input('Enter a number:  '))
print(factorial(5))
print(factorial(num))
```

#Fibonacci series:                  
0,1,1,2,3,5,8,13,........

if n=0 0,
if n=1, 1 otherwise
f(n-1)+f(n-2)


```python
#fibonacci series
def fibonacci(n):
  if n==0:
    return 0
  elif n==1:
    return 1
  else:
    return fibonacci(n-1)+fibonacci(n-2)

print(fibonacci(5))
print(fibonacci(8))

#generate a fibonacci series upto 5 terms
```

#Sum of first n natural number

S(n)= 0 if n=0 [base case]

 =n+S(n-1) if n>0 [recursive case]

```python
#sum of first n natural numberc
def sum_natural_number(n):
  if n==0:
    return 0
  else:
    return n+ sum_natural_number(n-1)

print(sum_natural_number(5))
```

```python
#function to check if a string is palindrome

#[::-1] to reverse a string

def check_palindrome(stri):
  y=stri[::-1]
  if stri==y:
    print("Palindrome")
  else:
    print("Not palindrome")

check_palindrome("aba")
```

```python
#function to count how many vowels in a string
def vowel_count(str):
  vowels="aeiouAEIOU"
  count=0
  for char in str:
    if char in vowels:
     count+=1
  return count
  #,"e","i","o","u"

vowel_count("applea")
```

```python
#function to remove all  duplicate elements from a list

def duplicate_remover(list1):
  unique_list=set()
  for i in list1:
    if i not in unique_list:
      unique_list.add(i)
  print(list(unique_list))

#or
def duplicate_remove(list1):
  for i in list1:
    if list1.count(i)>1:
      list1.remove(i)
  print(list1)


list1=[1,2,3,3,4,4,5]
duplicate_remover(list1)

duplicate_remove(list1)
```

Armstrong number: 153= 1^3+5^3+3^3


```python
#check whether armstrong or not
import builtins
def armstrong_check(num):
  digits=str(num)
  power=len(digits)
  total=builtins.sum(int(d) ** power for d in digits)
  return num==total

num=int(input('Enter a number: '))
if armstrong_check(num):
  print(num,'is an armstrong number')
else:
  print(num,'is not an armstrong number')
```

```python
def armstrong_check(num):
    digits = str(num)
    power = len(digits)
    total = sum(int(digit) ** power for digit in digits)
    return num == total
```

```python
#palindrome
def is_palindrome(num):
  x=str(num)
  y=x[::-1]

  if x == y:
    print("Palindrome")
  else:
    print("Not palindrome")

is_palindrome(121)
is_palindrome(12)
```

```python
#find maximum element in a list
num_list=[101,1,2,3,5,100]

def find_max(num):
  maximum=num[0]
  for i in num:
    if i>maximum:
      maximum=i
  return maximum

find_max(num_list)
```
