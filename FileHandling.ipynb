#File Handling



```python
#read-r
filename='/content/file.txt'
with open(filename,'r') as file:
  content=file.read()
print(content)
```

```python
#read line by line
filename='/content/file.txt'
with open(filename,'r') as file:
  for line in file:
    print(line)

```

```python
#print content of file as list
filename='/content/file.txt'
with open(filename,'r') as file:
  content=file.readlines()
print(content)
```

```python
#write mode
filename='/content/file.txt'
with open(filename,'w') as file:
  file.write('here is another line')

print(f'file{filename} has been written successfully')

```

```python
#append mode
filename='/content/file.txt'
with open(filename,'a') as file:
  file.write('here is another line')

print(f'file{filename} has been written successfully')
print(file)
```

```python
filename='/content/file.txt'
with open(filename,'r') as file:
   content=file.read()
   print(content)
```

```python
#create mode- x
filename='/content/demo100.txt'
try:
  with open(filename,'x') as file:
    file.write('this file is created using x mode')
    print(f'file {filename} created successfully')
except FileExistsError:
    print(f'error: file {filename} already exits')
```

#Combined modes (Read and Write/Append)

r+(read and write)

w+(write and read)

a+(append and read)

```python
filename='/content/file.txt'
with open(filename,'w+') as file:
  file.write("Hello word")
  content=file.read()
  file.seek(0) #move the cursor back at the beginning
print(content)
```

```python
filename='/content/file.txt'
with open(filename,'w+') as file:
  file.write("Hello world")
  file.seek(0) #move the cursor back at the beginning
  content=file.read()
print(content)
```
