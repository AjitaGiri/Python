#Pillars of OOP
ENCAPSULATION: wrapping/bundling data and method into a single unit (class)    
This can be done by making data private and exposing data only through public methods (getter/setter)


```python
#Encapsulation
class BankAccount:
  def __init__(self,owner,balance):
    self.owner=owner
    self.__balance=balance

#getter method
  def get_balance(self):
    return self.__balance

#setter method
  def  deposit(self,amount):
    if amount>0:
      self.__balance+=amount
    else:
      print("Deposit must be positive")

  def withdraw(self,amount):
      if 0<amount<self.__balance:
        self.__balance-=amount
      else:
        print('Insufficient balance or invalid amount')

b1=BankAccount('Jake',10000)
print(b1.owner)
b1.get_balance()
b1.deposit(5000)
b1.withdraw(500)
b1.get_balance()
```

#Abstraction
ABSTRACTION: data hiding....hiding implementation details from user and showing only necessary features of an object

In python it can be achieved by:          
-> abstract classes or interface          
-> Users interact with the interface not internalimplementation               
-> Great for reducing complexity and  increases flexibility

In python we can achieve this by *abc* module

```python
from abc import ABC, abstractmethod  #ABC: ABSTRACT BASE CLASS
#abstract class
class Vehicle(ABC):
 @abstractmethod #decorator
 def start(self):
   pass

 def stop(self):
  pass

#concrete class
class Car(Vehicle):
  def start(self):
    print('Car engine starts with a key')

  def stop(self):
    print('Car engine stops when you take out key')

c=Car()
c.start()
c.stop()
```

```python
#Methods
'''
  instance method: common type of method
  operates on intance of class
  automatically takes self as the first parameter
  through self, instance methods can access and modify the instance attributes
'''
'''
  Static method:
  method inside a class that doesnt take self as parameter
  use @static decorator
  Used when we dont have to access or modify the instance data or method
'''

'''
   Class method:
   cls as first parameter


'''
```

```python
#static method
class Student:
  @staticmethod
  def college_name():
    print('TU University')

  def __init__(self,name):
    self.name=name

s1=Student('Jake')
s2=Student('Sunoo')
s1.college_name()
s2.college_name()

```

```python
#class method
class Student:
  college_name='TU'

  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def details(self):
    return f"{self.name},marks:{self.marks} college:{self.college_name}"

  @classmethod
  def change_college(cls,new_name):
    cls.college_name=new_name


s1=Student('Jake',100)
s1.details()
s1.change_college('KU')
s1.details()
s1.college_name
s1.college_name='Pu'
print(s1.college_name)
s1.details()
```

#Inheritance
new /derived class is created from existing /base /parent class

child class can access the attributes and property of parent class.

super() access the methods of parent class




```python
#Inheritance
class Car:
  color="red"
  @staticmethod
  def start_car():
    print("CAR has started......")

  @staticmethod
  def stop_car():
    print("CAR has stopped.....")

class Toyota(Car):
  def __init__(self,model):
    self.model=model

t1=Toyota('LandRover')
t1.start_car()
t1.stop_car()
print(t1.model)
print(t1.color)
```

#Types of Inheritance
1. Single
2. Multilevel: parent, child, grandchild
3. Multiple: new class is derived from two or more parent class
4. Hierarchical: one parent class, and from that class multiple child class are created


```python
#Hierarchical Inheritance

class Car:
  color="red"
  @staticmethod
  def start_car():
    print("CAR has started......")

  @staticmethod
  def stop_car():
    print("CAR has stopped.....")

class Toyota(Car):
  def __init__(self,model):
    self.model=model

class Hyundai(Car):
  def __init__(self,speed):
    self.speed=speed

class Tata(Car):
  def __init__(self,horsepower):
    self.horsepower=horsepower

t1=Tata(2500)
t1.start_car()

h1=Hyundai(6000)
h1.speed
h1.stop_car()
```

```python
#Multilevel Inheritance
class Animal:
  def eat(self):
    print("This animal can eat")

class Mammal(Animal):
  def walk(self):
    print("This mammal can walk")

class Dog(Mammal):
  def bark(self):
    print("Dog can bark")

d1=Dog()
d1.walk()
d1.bark()
d1.eat()
```

```python
#Multiple Inheritance
class Mammal:
  def give_birth(self):
    print("Can give birth....")

class Winged_Animal:
  def fly(self):
    print("Can fly....")

class Bat(Mammal, Winged_Animal):
  pass

b1=Bat()
b1.give_birth()
b1.fly()
```

```python
#robot with vision +language abilities

class VisionSystem:
  def process_image(self, image):
    print(f'visionsystem: processing image {image}')

  def detect_objects(self):
      print('Vision system: objects detected in the scence')

class LanguageSystem:
  def process_text(self,text):
    print(f'Language system: Processing text {text}')

  def translate(self,text,target_lang='English'):
    print(f'Language system: Translating {text} into {target_lang}')

class Robot(VisionSystem,LanguageSystem):
  def __init__(self,name):
    self.name=name

  def introduce(self):
    print(f'Hello I am {self.name}, an AI powered robot')

  def multi_task(self,image,text):
    print('\n--------- Robot Multi-tasking------')
    self.process_image(image)
    self.detect_objects()
    self.process_text(text)
    self.translate(text,'French')

r1=Robot('Atlas')
r1.introduce()
r1.multi_task('street_view.jpg','Hello_World')

```

#super()
it is used to access the method of parent class.

```python
class Parent:
  def __init__(self,name):
    self.name=name
    print(f'Parent constructor called for {self.name}')

class Child(Parent):
  def __init__(self, name,age):
    super().__init__(name)
    self.age=age
    print(f'Child constructor called for {self.name}, age {self.age}')

c1=Child('Jake',23)

```

#Polymerphism
having many forms

same name but different functionality

example: Operator overloading

```python
class Dog:
  def sound(self):
    return 'Bark'

class Cat:
  def sound(self):
    return 'Meow'

for animal in [Dog(), Cat()]:
  print(animal.sound())
```

```python
class Animal:
  def sound(self):
    return 'Some sound'

class Dog(Animal):
  def sound(self):
    return 'bark'

class Cat(Animal):
  def sound(self):
    return 'Meow'

animals=[Animal(),Dog(),Cat()]
for a  in animals:
  print(a.sound())
```
